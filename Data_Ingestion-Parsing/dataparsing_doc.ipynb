```python
from langchain_community.document_loaders import Docx2txtLoader, UnstructuredWordDocumentLoader
```

```python
#method -1: Using Docx2txtLoader
try:
    docx_loader = Docx2txtLoader("data\Doc\📑 land registration process.docx")
    docs=docx_loader.load()
    print(f"Documents loaded: {len(docs)}")
    print(f"page content:{docs[0].page_content[:100]}")  
except Exception as e:
    print(f"An error occurred while loading the document using Docx2txtLoader: {e}")
```

```python
#method-2: Using UnstructuredWordDocumentLoader

try:
    docs =UnstructuredWordDocumentLoader("data\Doc\📑 land registration process.docx",mode="elements")
    unstructure_docs = docs.load()
    print(f"length of unstructure_docs {len(unstructure_docs)} elements")
    
   # for i,docs in  enumerate(unstructure_docs):
       # print(f"element {i+1}:")
       # print(f"page content:{docs.page_content[:50]}")
        
except Exception as e:
    print(f"error{e}")
```
