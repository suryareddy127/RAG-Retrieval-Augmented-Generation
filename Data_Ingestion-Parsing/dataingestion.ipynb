## Text Splitters
### Method-1: Character Text Splitter

```python
# method-1 character text splitter
text=documents[0].page_content
text 
```

```python
char_splitter = CharacterTextSplitter(
    separator="\n",     # splitting at new lines
    chunk_size=500,     # size of each chunk
    chunk_overlap=50,   # number of characters to overlap between chunks
    length_function=len # function to measure length
)
char_chunks = char_splitter.split_text(text)
print(f"number of character chunks: {len(char_chunks)}")
print(f"first chunk:\n{char_chunks[0][:500]}...")
```

```python
print(len(char_chunks))
print(char_chunks[1])
print("-----------------------------------------")
print(char_chunks[7])
```

### method -2 Recursive Text Splitter

```python
#method -2  recursive character text splitter (most recommended)
# RecursiveCharacterTextSplitter 
recursive_splitter = RecursiveCharacterTextSplitter(
    separators=["\n\n", "\n", " ", ""],
    chunk_size=500,
    chunk_overlap=50,
    length_function=len
)
# print(recursive_splitter)
recursive_chunks = recursive_splitter.split_text(text)
print(f"number of recursive chunks: {len(recursive_chunks)}")
print(f"first recursive chunk:\n{recursive_chunks[0][:500]}...")
```

```python
print(recursive_chunks[1])
print("-----------------------------------------")
print(recursive_chunks[2])
print("-----------------------------------------")
print(recursive_chunks[3])
print("-----------------------------------------")
```

```python
# RecursiveCharacterTextSplitter example 2
sample_texxt = "It is an AI framework that enhances large language models (LLMs) by integrating them with external knowledge sources to produce more accurate, up-to-date, and contextually relevant responses. Instead of relying solely on their fixed training data, RAG systems first retrieve pertinent information from external databases, documents, or knowledge bases using techniques like semantic search, keyword search, or vector search, before generating an answer."

splitter = RecursiveCharacterTextSplitter(
    separators=[" "],
    chunk_size=100,
    chunk_overlap=25,
    length_function=len
)
chunks = splitter.split_text(sample_texxt)
print(f"number of chunks: {len(chunks)}")

for i in range(len(chunks) - 1):
    print(f"chunk{i+1}: '{chunks[i]}'")
    print(f"chunk{i+2}: '{chunks[i+1]}'")
    
    print()
```

```python
#method-3  token text splitter
token_splitter = TokenTextSplitter(
    chunk_size=100,
    chunk_overlap=20
)
#print(token_splitter)
token_chunks = token_splitter.split_text(text)
print(f"number of token chunks: {len(token_chunks)}")

print(f"first token chunk: {token_chunks[0][:100]}...")
```
